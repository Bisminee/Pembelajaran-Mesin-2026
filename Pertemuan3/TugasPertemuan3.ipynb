```python
import pandas as pd

df = pd.read_csv('data/122KBwbc.csv')
df.head()
```

```python
# 1. Pisahkan variabel yang dapat dan tidak dapat digunakan

# Kolom yang tidak dapat digunakan: id (identifier) dan Unnamed: 32 (kolom kosong)
unuse = ['id', 'Unnamed: 32']

# Kolom target yang akan diprediksi
target = 'diagnosis'

# Kolom fitur numerik yang dapat digunakan
usable = [c for c in df.columns if c not in unuse + [target]]

# Buang kolom yang tidak dapat digunakan
df = df.drop(columns=unuse)

print(f'Variabel tidak dapat digunakan : {unuse}')
print(f'Kolom target                   : {target}')
print(f'Jumlah variabel dapat digunakan: {len(usable)}')
print(f'Variabel dapat digunakan       : {usable}')
```

```python
# 2. Encoding kolom target 'diagnosis' (B = 0, M = 1)
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])

print(f'Kelas label: {list(le.classes_)} -> {list(le.transform(le.classes_))}')
df[['diagnosis']].head()
```

```python
# 3. Standarisasi semua kolom numerik
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[usable] = scaler.fit_transform(df[usable])

# Cek hasil standardisasi (mean ~ 0, std ~ 1)
df[usable].describe().loc[['mean', 'std']]
```

```python
# 4. Stratified split data latih dan data uji (80:20)
from sklearn.model_selection import train_test_split

X = df[usable]
y = df[target]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0, stratify=y
)

print(f'Jumlah data asli : {df.shape[0]}')
print(f'Jumlah data train: {X_train.shape[0]}')
print(f'Jumlah data test : {X_test.shape[0]}')
print('=== Proporsi label ===')
print(f'Data asli :\n{y.value_counts(normalize=True)}')
print(f'Data train:\n{y_train.value_counts(normalize=True)}')
print(f'Data test :\n{y_test.value_counts(normalize=True)}')
```
