# Tugas Praktikum - Wisconsin Breast Cancer
Dataset berisi 569 data untuk mendiagnosis kanker **Malignant (M)** dan **Benign (B)**.

Tahapan:
1. Memisahkan variabel yang dapat dan tidak dapat digunakan.
2. Encoding pada kolom `diagnosis`.
3. Standardisasi pada semua kolom numerik.

## Persiapan

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Memuat Data

```python
df = pd.read_csv('wbc.csv')
df.info()
df.head()
```

## 1. Memisahkan Variabel 

- **Tidak dapat digunakan**: `id` (hanya identitas unik, tidak punya nilai prediktif) dan `Unnamed: 32` (kolom kosong/semua NaN akibat trailing comma pada CSV).
- **Target (label)**: `diagnosis` (M/B).
- **Dapat digunakan**: seluruh kolom numerik fitur (30 kolom).

```python
# Kolom yang tidak dapat digunakan
unusable_cols = ['id', 'Unnamed: 32']

# Kolom target yang akan diprediksi
target_col = 'diagnosis'

# Kolom yang dapat digunakan (fitur numerik)
usable_cols = [c for c in df.columns if c not in unusable_cols + [target_col]]

# Buang kolom yang tidak digunakan
df = df.drop(columns=unusable_cols)

print('Variabel tidak dapat digunakan :', unusable_cols)
print('Kolom target                   :', target_col)
print('Jumlah variabel dapat digunakan:', len(usable_cols))
print('Variabel dapat digunakan       :', usable_cols)
```

## 2. Encoding Kolom diagnosis

```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])

# Lihat hasil mapping
print('Mapping kelas:', dict(zip(le.classes_, le.transform(le.classes_))))
df['diagnosis'].value_counts()
```

## 3. Standardisasi Semua Kolom Numerik

Standardisasi mengubah data menjadi **z-score**: `z = (x - mean) / std`,
sehingga rata-rata ≈ 0 dan standar deviasi ≈ 1. Dilakukan pada seluruh kolom fitur numerik.

```python
scaler = StandardScaler()
df[usable_cols] = scaler.fit_transform(df[usable_cols])

df.head()
```

## Verifikasi

```python
# Cek hasil standardisasi: mean ~ 0, std ~ 1
df[usable_cols].describe().loc[['mean', 'std']]
```
